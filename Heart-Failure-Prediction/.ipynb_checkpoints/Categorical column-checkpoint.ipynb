```python
import pandas as pd
import seaborn as sns
import numpy as np 
df=pd.read_csv('heart.csv')
```

```python
df
```

 ## Distribution Plot for Numeric Columns

```python
sns.displot(df['RestingBP'],kde=False,color='blue')
```

```python
sns.distplot(df['Cholesterol'],kde=True,color='blue')
```

```python
sns.distplot(df['MaxHR'],kde=True,color='blue')
```

```python
sns.distplot(df['Oldpeak'],kde=True,color='blue')
```

## Pie Charts for Categorical 

```python
df.groupby('Sex').size()
```

```python
df.groupby('Sex').size().plot(kind='pie',autopct='%.0f',label="Male and female")
```

```python
df.groupby('ChestPainType').size()
```

```python
df.groupby('ChestPainType').size().plot(kind='pie',autopct='%.0f')
```

```python
df.groupby('ChestPainType').size().plot(kind='pie',autopct='%.1f')
```

```python
df.groupby('ST_Slope').size().plot(kind='pie',autopct='%.0f')
```

```python
df.groupby('HeartDisease').size().plot(kind='pie',autopct='%.0f')
```
